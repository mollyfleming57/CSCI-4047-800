3. For *Income*, *Education*, and *Prestige* calculate and display the following data:
    - Minimum Value
    - 1st Quartile
    - Median
    - Mean
    - 3rd Quartile
    - Maximum Value

```python
import numpy as np

min = np.min(df.prestige) 
q1 = np.percentile(df.prestige, 25)
med = np.median(df.prestige)
q3 = np.percentile(df.prestige, 75)
max = np.max(df.prestige)

print('------------ FNS prestige -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Q3: {q3}")
print(f"Max: {max}")
```

```python
import numpy as np

min = np.min(df.education) 
q1 = np.percentile(df.education, 25)
med = np.median(df.education)
q3 = np.percentile(df.education, 75)
max = np.max(df.education)

print('------------ FNS education -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Q3: {q3}")
print(f"Max: {max}")
```

```python
import numpy as np

min = np.min(df.income) 
q1 = np.percentile(df.income, 25)
med = np.median(df.income)
q3 = np.percentile(df.income, 75)
max = np.max(df.income)

print('------------ FNS income -------------')
print(f"Min: {min}")
print(f"Q1: {q1}")
print(f"Median: {med}")
print(f"Q3: {q3}")
print(f"Max: {max}")


```

8. Calculate and display the *variance* and *standard deviation* for **income**


```python
standard_dev = df['income'].std()
print()
print(f"Standard Deviation:  {standard_dev:.2f}")

variance = standard_dev ** 2
print(f"Variance:            {variance:.2f}")
```
