```python
# Prerequisites
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay, classification_report, recall_score

```

## Step 1 - Data exploration

Load your data and get to know it: how many rows and columns, what the columns mean, and
how balanced the classes are. Class balance matters: if 90% of rows are one class, a model
can look accurate just by guessing that class every time.

```python
# GIVEN: load the dataset. Change this path to your group's file when you have it.
#NASA data used
df = pd.read_csv("nasa.csv")

# Set this to the name of the column you are predicting.
target = "Hazardous"

df.head()
```

```python
# TODO: explore the data
print("Shape (rows, columns):", df.shape)
print()
df.info()

print()

print(df[target].value_counts())
print()
print(df[target].value_counts(normalize=True).round(3))

print()

baseline_all = df[target].value_counts(normalize=True).max()
print(f"\nBaseline: always guessing 'not hazardous' is already {baseline_all:.1%} accurate.")

print()

print("Total missing values:", df.isna().sum().sum())
print()
print(df["Orbiting Body"].value_counts())
print()
print(df["Equinox"].value_counts())

print()

df[["Absolute Magnitude", "Est Dia in KM(max)", "Relative Velocity km per sec",
    "Miss Dist.(Astronomical)", "Minimum Orbit Intersection", "Orbital Period"]].describe()

print()

print(df[["Absolute Magnitude", "Est Dia in KM(max)", "Relative Velocity km per sec",
    "Miss Dist.(Astronomical)", "Minimum Orbit Intersection", "Orbital Period"]].describe())
```

## Step 2 - Preprocessing

Get the data ready for the models. You will typically need to:

- Drop columns that are just identifiers or are not useful (e.g. an ID column).
- Handle missing values (fill numeric columns with the median, text columns with the mode).
- Encode categorical (text) columns into numbers with `pd.get_dummies(...)` or `LabelEncoder`.
- Separate the target `y` from the features `X`, then split into train and test sets.

```python
# TODO: drop unhelpful columns and handle missing values
data = df.copy()

drop_cols = [
    "Neo Reference ID", "Name", "Orbit ID",
    "Orbiting Body", "Equinox",
    "Close Approach Date", "Epoch Date Close Approach", "Orbit Determination Date",
    "Epoch Osculation", "Perihelion Time",
    "Est Dia in KM(min)", "Est Dia in M(min)", "Est Dia in M(max)",
    "Est Dia in Miles(min)", "Est Dia in Miles(max)",
    "Est Dia in Feet(min)", "Est Dia in Feet(max)",
    "Relative Velocity km per hr", "Miles per hour",
    "Miss Dist.(lunar)", "Miss Dist.(kilometers)", "Miss Dist.(miles)",
]
data = data.drop(columns=drop_cols)

print("Columns left:", data.shape[1])
print(list(data.columns))
```

```python
# TODO: build X (features) and y (target), then encode the categorical columns
y = data[target].map({False: 0, True: 1})

X = data.drop(columns=[target])

print("X shape:", X.shape)
print(y.value_counts())
```

```python
# TODO: split into training and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1, stratify=y)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("Train:", X_train.shape, "| Test:", X_test.shape)
```

## Step 3 - Build the models

Each classifier is only a few lines: create it, then call `.fit(X_train, y_train)`.

```python
# TODO: train a Decision Tree
dtree = DecisionTreeClassifier(max_depth=5, random_state=1)
dtree.fit(X_train, y_train)

importances = pd.Series(dtree.feature_importances_, index=X.columns).sort_values(ascending=False)
print(importances.head(8).round(3))

```

```python
# TODO: train a Naive Bayes classifier
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
print("Naive Bayes trained.")
```

## Step 4 - Evaluate and compare

For **each** model, predict on the test set and report:

- accuracy (`accuracy_score`)
- the confusion matrix (`ConfusionMatrixDisplay.from_predictions`)
- precision, recall, and F1 (`classification_report`)

Then compare: which model is more accurate, and what do the precision and recall for the
class you care about tell you?

```python
# TODO: evaluate the Decision Tree
dt_pred = dtree.predict(X_test)
print("Decision Tree accuracy:", round(accuracy_score(y_test, dt_pred), 4))

ConfusionMatrixDisplay.from_predictions(y_test, dt_pred, cmap="Blues",
                                        display_labels=["Not hazardous", "Hazardous"])
plt.title("Decision Tree")
plt.show()

print(classification_report(y_test, dt_pred, target_names=["Not hazardous", "Hazardous"]))
```

```python
# TODO: evaluate the Naive Bayes model (same steps as above)
nb_pred = nb_model.predict(X_test)
print("Naive Bayes accuracy:", round(accuracy_score(y_test, nb_pred), 4))

ConfusionMatrixDisplay.from_predictions(y_test, nb_pred, cmap="Oranges",
                                        display_labels=["Not hazardous", "Hazardous"])
plt.title("Naive Bayes")
plt.show()

print(classification_report(y_test, nb_pred, target_names=["Not hazardous", "Hazardous"]))

```

```python
# TODO: compare the two models' accuracy against a baseline
proba = nb_model.predict_proba(X_test[:5])

for p in proba:
    print(f"Not hazardous: {p[0]:.2f} | Hazardous: {p[1]:.2f}")

print("Baseline accuracy:", round(baseline_all, 3))
print("Decision Tree accuracy:", round(accuracy_score(y_test, dt_pred), 3))
print("Naive Bayes accuracy:", round(accuracy_score(y_test, nb_pred), 3))
print()
print("Decision Tree hazardous recall:", round(recall_score(y_test, dt_pred), 3))
print("Naive Bayes hazardous recall:", round(recall_score(y_test, nb_pred), 3))

```
